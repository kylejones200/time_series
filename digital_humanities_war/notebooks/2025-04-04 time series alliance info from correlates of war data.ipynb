```python
import pandas as pd

# Read the dyad dataset
df_dyad = pd.read_csv("/Users/kylejonespatricia/Downloads/version4.1_csv/alliance_v4.1_by_dyad.csv")

def convert_date(day, month, year):
    """Convert day, month, year to a Timestamp, or return NaT if any part is missing."""
    try:
        # If any component is missing, return NaT
        if pd.isna(day) or pd.isna(month) or pd.isna(year):
            return pd.NaT
        return pd.Timestamp(int(year), int(month), int(day))
    except Exception as e:
        return pd.NaT

# Create start_date and end_date columns
df_dyad['start_date'] = df_dyad.apply(
    lambda row: convert_date(row['dyad_st_day'], row['dyad_st_month'], row['dyad_st_year']),
    axis=1
)
df_dyad['end_date'] = df_dyad.apply(
    lambda row: convert_date(row['dyad_end_day'], row['dyad_end_month'], row['dyad_end_year']),
    axis=1
)

# Calculate the duration in days (if both dates are available)
df_dyad['duration'] = (df_dyad['end_date'] - df_dyad['start_date']).dt.days

# Display the first few rows to verify the new columns
print(df_dyad[['start_date', 'end_date', 'duration']].head())
```

```python
# Fill missing end_date values with April 1, 2013
df_dyad["end_date"] = df_dyad["end_date"].fillna(pd.Timestamp("2013-04-01"))

# Recalculate duration if necessary
df_dyad["duration"] = (df_dyad["end_date"] - df_dyad["start_date"]).dt.days

# Display a sample of the updated DataFrame
print(df_dyad[['start_date', 'end_date', 'duration']].head())
```
