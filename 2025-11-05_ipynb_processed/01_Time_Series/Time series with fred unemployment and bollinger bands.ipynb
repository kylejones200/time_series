```python
import pandas as pdimport pandas_datareader.data as webimport datetimeimport matplotlib.pyplot as pltfrom prophet import Prophetstart = datetime.datetime(2010, 1, 1)end = datetime.datetime(2024, 10, 20)df = web.DataReader('unrate', 'fred', start, end)df.reset_index(inplace=True)df.columns = ["ds", "y"]df.dropna(inplace=True)model = Prophet()model.fit(df)future = model.make_future_dataframe(periods=12, freq='MS')forecast = model.predict(future)fig, ax = plt.subplots(figsize=(12, 6))ax.plot(df['ds'], df['y'], 'ko', markersize=3, alpha=0.6, label='Actual')ax.plot(forecast['ds'], forecast['yhat'], 'b-', linewidth=1.5, label='Forecast')ax.fill_between(forecast['ds'], forecast['yhat_lower'], forecast['yhat_upper'],                 color='b', alpha=0.15)ax.set_xlabel('Date')ax.set_ylabel('Unemployment Rate')ax.legend(frameon=False, loc='best')ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()df1 = web.DataReader('DHHNGSP', 'fred', start, end)df1.reset_index(inplace=True)df1.columns = ["ds", "y"]m = Prophet()m.fit(df1)future1 = m.make_future_dataframe(periods=12, freq='MS')fcst1 = m.predict(future1)fig, ax = plt.subplots(figsize=(12, 6))ax.plot(df1['ds'], df1['y'], 'ko', markersize=3, alpha=0.6, label='Actual')ax.plot(fcst1['ds'], fcst1['yhat'], 'r-', linewidth=1.5, label='Forecast')ax.fill_between(fcst1['ds'], fcst1['yhat_lower'], fcst1['yhat_upper'],                 color='r', alpha=0.15)ax.set_xlabel('Date')ax.set_ylabel('Price')ax.legend(frameon=False, loc='best')ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show() def bollinger_bands(df, target_col='adjClose'):    df = df.dropna().copy()    df['20 Day MA'] = df[target_col].rolling(20).mean()    df['20 Day MA_lower'] = df['20 Day MA'] - df[target_col].rolling(20).std() * 2    df['20 Day MA_upper'] = df['20 Day MA'] + df[target_col].rolling(20).std() * 2    return dfstart2 = datetime.datetime(2024, 4, 1)end2 = datetime.datetime(2024, 10, 20)df2 = web.DataReader('DHHNGSP', 'fred', start2, end2)df2["adjClose"] = df2["DHHNGSP"]df2 = bollinger_bands(df2.sort_values(by='DATE'))fig, ax = plt.subplots(figsize=(12, 6))ax.plot(df2.index, df2['adjClose'], 'k-', linewidth=1.5, label='Price')ax.plot(df2.index, df2['20 Day MA'], 'b-', linewidth=1.5, label='MA')ax.fill_between(df2.index, df2['20 Day MA_lower'], df2['20 Day MA_upper'],                 color='b', alpha=0.2)ax.set_xlabel('Date')ax.set_ylabel('Price (USD)')ax.legend(frameon=False, loc='best')ax.spines['top'].set_visible(False)ax.spines['right'].set_visible(False)plt.tight_layout()plt.show()
```
