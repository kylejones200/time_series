```python
import pandas as pdimport numpy as npimport matplotlib.pyplot as pltdate_range = pd.date_range(start="2023-01-01", periods=100, freq="D")df = pd.DataFrame({"value": np.random.randn(100).cumsum()}, index=date_range)df["lagged"] = df["value"].shift(1)df["rolling_mean"] = df["value"].rolling(window=7).mean()df["rolling_std"] = df["value"].rolling(window=7).std()df["weekly_sum"] = df["value"].resample("W").sum()df["weekly_sum_shifted"] = df["weekly_sum"].shift(1).fillna(0)df["weekly_avg"] = df["weekly_sum"].rolling(window=2).mean()fig, axes = plt.subplots(2, 2, figsize=(12, 8))axes[0, 0].plot(df.index, df["value"], linewidth=1.5, color='k')axes[0, 0].set_title("Original Series")axes[0, 1].plot(df.index, df["rolling_mean"], linewidth=1.5, color='b', label='Mean')axes[0, 1].fill_between(df.index, df["rolling_mean"] - df["rolling_std"],                         df["rolling_mean"] + df["rolling_std"], alpha=0.2, color='b')axes[0, 1].set_title("Rolling Statistics")axes[1, 0].plot(df["weekly_sum"].index, df["weekly_sum"], marker='o',                 markersize=3, linewidth=1.5, color='g')axes[1, 0].set_title("Weekly Aggregation")axes[1, 1].plot(df.index, df["value"], linewidth=1, color='k', alpha=0.5, label='Original')axes[1, 1].plot(df.index, df["lagged"], linewidth=1.5, color='r', label='Lagged')axes[1, 1].set_title("Lagged Features")axes[1, 1].legend(frameon=False)for ax in axes.flat:    ax.spines['top'].set_visible(False)    ax.spines['right'].set_visible(False)    ax.tick_params(axis='x', rotation=45)plt.tight_layout()plt.show()
```
